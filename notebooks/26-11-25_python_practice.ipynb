Question 1: Task: A listing is a duplicate if another listing has the same company_id, title and description. Count how many companies have at least one duplicate listing. Return a one-row DataFrame with the column duplicate_companies.

```python
import pandas as pd

job_listings = pd.DataFrame({
    "job_id": range(1, 10),
    "company_id": [827, 827, 827, 153, 153, 153, 100, 100, 200],
    "title": ["Data Analyst", "Data Analyst", "BI Developer", "Data Analyst", "Data Analyst",
              "Data Analyst", "Engineer", "Analyst", "Analyst"],
    "description": ["SQL", "SQL", "Power BI", "Python", "Python", "Python", "Java", "SQL", "SQL"],
})
```

```python
job_listings['si_no'] = job_listings.groupby(['company_id', 'title', 'description'])['job_id'].rank(method = 'first')

dup_companies = job_listings.loc[job_listings['si_no'] > 1, 'company_id'].drop_duplicates()

result = pd.DataFrame({'duplicate_companies': [len(dup_companies)]})

result
```

Question 2: Return the average star rating for each product in each month, rounded to 2 decimals. Columns: month (as a number), product_id, avg_stars. Order by month, then product_id.

```python
reviews = pd.DataFrame({
    "review_id": range(1, 9),
    "user_id": [123, 265, 362, 192, 981, 111, 222, 333],
    "submit_date": pd.to_datetime(["2022-06-08", "2022-06-10", "2022-06-18", "2022-07-26",
                                   "2022-07-05", "2022-06-25", "2022-07-30", "2022-07-01"]),
    "product_id": [50001, 50001, 50001, 50001, 50002, 50002, 50001, 50002],
    "stars": [4, 4, 3, 3, 2, 5, 5, 4],
})
```

```python
reviews['month'] = reviews['submit_date'].dt.month
reviews['year'] = reviews['submit_date'].dt.year

reviews_agg = reviews.groupby(['month', 'product_id'])['stars'].mean().reset_index(name = 'avg_stars')
# reviews_agg = reviews.groupby(['month', 'product_id']).agg(avg_stars = ('stars', 'mean')).reset_index()

reviews_agg['avg_stars'] = reviews_agg['avg_stars'].round(2)

reviews_agg.sort_values(['month', 'product_id'])
```

Question 3: Return the customer_id of every customer who bought all products in products, ordered ascending. Customer 3 bought product 30 twice, so watch for duplicates.

```python
customer_contracts = pd.DataFrame({
    "customer_id": [1, 1, 1, 2, 2, 3, 3, 3, 3, 4],
    "product_id":  [10, 20, 30, 10, 20, 10, 20, 30, 30, 30],
})
products = pd.DataFrame({"product_id": [10, 20, 30]})
```

```python
merged = pd.merge(products, customer_contracts, on = 'product_id', how = 'left').sort_values('customer_id')

merged_agg = merged.groupby('customer_id').agg(counter = ('product_id', 'nunique'))

merged_agg = merged_agg.loc[merged_agg['counter'] == 3].reset_index()

# n_products = products['product_id'].nunique()
# merged_agg = merged_agg.loc[merged_agg['counter'] == n_products].reset_index()

merged_agg[['customer_id']]
```
